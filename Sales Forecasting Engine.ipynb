# Weekly Sales Report For XLNT Co.

##  Sales Forecast for Wireless Mouse

```python
# Task 1 - Creating the variables
product_name = "Wireless Mouse"
selling_price_per_unit = 12500
mouse_price = selling_price_per_unit #redefining for ease when answering the bonus challenge questions
units_sold = 180
weekly_sales_target = 200
forecast_growth_rate = 15
current_stock = 250
minimum_stock_level = 120
product_available = True

# Task 2 - Calculate total weekly revenue
weekly_revenue = selling_price_per_unit * units_sold

# Task 3- Forecast Next Week's Sales
forecast_sales = int(units_sold + (units_sold * forecast_growth_rate/100))

# Task 4 - Forecast Next week's revenue
forecast_revenue = selling_price_per_unit * forecast_sales

# Task 5 - Update current stock
current_stock -= units_sold

# Task 6 - Find out if sales target was met
target_achieved = units_sold >= weekly_sales_target

# Task 7 - Determine if restocking is required
need_for_restock = current_stock < minimum_stock_level

# Task 8 - Find out if product is readytion
promotion_criteria = product_available and target_achieved

# Task 9 - Determine if demand is high
demand_state = forecast_sales > 200 or forecast_revenue > 2500000

# Task 10 - Generate weekly sales report
print("="*50)
print("XLNT CO. WEEKLY SALES REPORT".center(50))
print("="*50)
print()
print(f"""Product Name: {product_name}

Selling Price: ₦{selling_price_per_unit:,}

Units Sold This Week: {units_sold}

Weekly Revenue: ₦{weekly_revenue:,}

Forecast Growth Rate: {forecast_growth_rate}%

Forecast Sales: {forecast_sales} Units

Forecast Revenue: ₦{forecast_revenue:,}

Current Stock: {current_stock} Units

Sales Target Achieved: {target_achieved}

Restocking Required: {need_for_restock}

Product Available: {product_available}

Promotion Eligible: {promotion_criteria}

High Demand Forecast: {demand_state}""")
print()
print("="*50)
```
